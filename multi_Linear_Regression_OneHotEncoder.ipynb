```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(handle_unknown = "ignore")
```

```python
encoder
```

```python
encode1 = encoder.fit_transform(df[['Location']])
```

```python
encode1
```

```python
df1 = pd.DataFrame(encode1.toarray(),columns = encoder.categories_[0])
```

```python
df1
```

```python
df = pd.concat([df1,df],axis =1)
```

```python
df
```

```python
df.drop(['Location'],axis = 1,inplace = True)
```

```python
df
```

```python
encode2 = encoder.fit_transform(df[['Property_Type']])
```

```python
encode2
```

```python
df2 = pd.DataFrame(encode2.toarray(),columns=encoder.categories_[0])
```

```python
df= pd.concat([df2,df],axis = 1)
```

```python
df.drop(['Property_Type'],axis = 1, inplace = True)
```

```python
df
```

```python
encode3 = encoder.fit_transform(df[['Furnishing']])
```

```python
encode3
```

```python
df3 = pd.DataFrame(encode3.toarray(),columns = encoder.categories_[0])
```

```python
df = pd.concat([df,df3],axis = 1)
```

```python
df.drop(['Furnishing'],axis = 1,inplace = True)
```

```python
df
```

```python
x = df.drop('House_Price_Lakhs',axis = 1)
y = df['House_Price_Lakhs']
```

```python
print("x shape:",x.shape)
print("y shape:",y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2)
```

```python
print(len(x_train))
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.predict(x_test)
```

```python
y_predict = model.predict(x_test)
```
